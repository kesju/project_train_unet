```python
import pandas as pd
from pathlib import Path

#   PARAMETRAI, APLANKAI  

# Duomenų aplankas su zive ir mit duomenimis
Duomenu_aplankas = Path.home() / 'DI/ZIVEO_2025/DUOMENYS_UPD'
Parametru_aplankas = Duomenu_aplankas/'PARAMETRAI'

noise_summary_file_name = "noise_summary_2.xlsx"
print(f"Reading noise summary file: {noise_summary_file_name}")

# Read the Excel file into a DataFrame
noise_summary_file__path = Parametru_aplankas / noise_summary_file_name
df = pd.read_excel(noise_summary_file__path)

# Show the first few rows to verify
# print(df.head())

# Per 'qlt' averages with fallback symbol
avg_per_qlt = {}
for q in [0, 1, 2]:
    values = df.loc[df['qlt'] == q, 'tp']
    if not values.empty:
        avg_per_qlt[q] = round(values.mean(), 1)
    else:
        avg_per_qlt[q] = '-'

# Overall average (not grouped)
overall_avg_tp = round(df['tp'].mean(), 1)

# Print results
print("\nAverage 'tp' per 'qlt':")
for qlt_val, avg in avg_per_qlt.items():
    print(f"qlt {qlt_val}: {avg}")

print(f"\nOverall average 'tp': {overall_avg_tp}")

```
