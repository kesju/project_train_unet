```python
# https://chatgpt.com/c/68ba9023-5d3c-8324-8f7c-d4d7f693bdca

# What it does:

# Extracts the trailing _1024_0_5_3_6 parts.
# Checks if both files have the same suffix.
# Verifies that "1024" matches SEGMENT_LENGTH.
# Verifies that "0_5" matches OVERLAP formatted with underscore.


# File names
model_file_name = "resunet_ecg_1024_0_5_3_6.keras"
noise_summary_basename = "noise_summary_1024_0_5_3_6"

# Configuration parameters (matched to training script)
CONFIG = {
    'FS': 200,              # Sampling rate
    'SEGMENT_LENGTH': 1024, # Segment length
    'OVERLAP': 0.5          # Overlap fraction
}

def check_consistency(model_file_name, noise_summary_basename, config):
    # --- Extract the common suffix part ---
    model_suffix = model_file_name.split("resunet_ecg")[-1].replace(".keras", "")
    noise_suffix = noise_summary_basename.split("noise_summary")[-1]
    
    if model_suffix != noise_suffix:
        raise ValueError(
            f"Suffix mismatch:\n"
            f"  model: {model_suffix}\n"
            f"  noise: {noise_suffix}"
        )

    # --- Check SEGMENT_LENGTH ---
    segment_str = str(config['SEGMENT_LENGTH'])
    if segment_str not in model_suffix:
        raise ValueError(
            f"SEGMENT_LENGTH mismatch: expected '{segment_str}' in suffix '{model_suffix}'"
        )

    # --- Check OVERLAP ---
    overlap_str = str(config['OVERLAP']).replace(".", "_")
    if overlap_str not in model_suffix:
        raise ValueError(
            f"OVERLAP mismatch: expected '{overlap_str}' in suffix '{model_suffix}'"
        )

    print("✅ All checks passed successfully!")
    print(f"  Suffix: {model_suffix}")
    print(f"  SEGMENT_LENGTH: {segment_str}")
    print(f"  OVERLAP: {overlap_str}")

if __name__ == "__main__":
    check_consistency(model_file_name, noise_summary_basename, CONFIG)
```
