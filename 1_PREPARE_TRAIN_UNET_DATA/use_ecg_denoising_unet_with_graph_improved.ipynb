```python
# Čia yra Python skriptas, skirtas EKG triukšmo aptikimui naudojant iš anksto apmokytą U-Net modelį
# su train_ecg_denoising_unetX.py. išskirčių (outliers) ir rspragų (rdropouts) neieškoma.
# Skriptas naudoja Keras bibliotekas, kad įkeltų modelį ir atliktų EKG signalų apdorojimą.

# Jei flag_plot_start_with_distortions_entire = True, išvedamas grafikas su anotuotais (dažoma geltonais)
# ir aptiktais triukšmais (dažoma raudonai)ir išsaugomas kaip paveikslėlis.
# R pikų nerodo.

# https://grok.com/chat/68e9c626-abf9-4fc3-9102-d82468902b6f

# Install missing dependencies for this notebook environment
%pip install -q keras tensorflow

import pandas as pd
from pathlib import Path
import os, math, sys

# Fix: Import keras directly for model loading
import keras


# === Išoriniai moduliai (lygiagretus aplankas) =================================
PARALLEL_PATH = Path().resolve().parent / "SUPL_FUNCTIONS"
sys.path.append(str(PARALLEL_PATH))

from project_util import find_project_root_by_name
from ecg_denoising_util import get_ecg_signal, get_ecg_noise_indices_annotated
from zive_util_ml import divide_signal_into_fragments
from use_ecg_denoising_util import run_ecg_denoising_pipeline
from use_ecg_denoising_util import ecg_filter
from ecg_denoising_util import plot_signal_write_plot_R_P_1

PROJECT_DIR = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
PROJECT_ROOT = find_project_root_by_name(target="PROJECT_TRAIN_UNET", start=PROJECT_DIR)
print("\nPROJECT ROOT DIR:", PROJECT_ROOT)
print("PROJECT DIR:", PROJECT_DIR)

        #   PARAMETRAI, APLANKAI IR KONFIGŪRACIJA 

# Duomenų aplankas su zive ir mit duomenimis
Duomenu_aplankas = PROJECT_ROOT/'DATA_ORIG'
MODEL_DIR = PROJECT_ROOT/'MODEL_UNET'

# Aplankas su EKG įrašais (.npy) ir anotacijomis (.json)
db_folder = 'ecg_zive_npy'  # zive
# db_folder = 'Igno'  # Igno Griskeviciaus

# Nuoroda į aplanką su EKG įrašais (.npy) ir anotacijomis (.json)
rec_dir = Path(Duomenu_aplankas, db_folder)


# start0
model_file_name = "resunet_ecg_1024_0_5_3_7.keras"

# resunet_ecg_1024_0_5_3_6.keras:
# 1024 - 'SEGMENT_LENGTH'
# 0_5 - 'OVERLAP'
# 3 - papildymas, leidžiantis susigaudyti su kokiais duomenimis buvo mokymas. 3 reiškia, kad naudojama mokymo imtis po 3-io
#  papildymo duomenimis (žr. 3-as etapas - duomenys mokymui). 
# 6 - pridėjau, kad būtų aišku, kad mokyta su train_ecg_denoising_unet6.py 


#  Configuration parameters (matched to training script)
CONFIG = {
    'FS': 200,              # Sampling rate
    'SEGMENT_LENGTH': 1024, # Segment length
    'OVERLAP': 0.5          # Overlap fraction
}

threshold = 0.15  # Residual threshold for noise detection (tune as needed)
threshold = 0.07  # Residual threshold for noise detection (tune as needed)
thresholds = [0.07, 0.08, 0.15]  # Residual thresholds for noise detection (tune as needed)
thresholds = [0.08]  # Residual threshold for noise detection (tune as needed)

segment_length = CONFIG['SEGMENT_LENGTH']
overlap = CONFIG['OVERLAP']
fs = CONFIG['FS']

# ECG įrašo filtravimui
fp = {  'type': 'lowpass',
            'method':'butterworth',
            'order':5,
            'sampling_rate':fs,
            'lowcut':0.5,
            'highcut':90 }

portion_length_in_secs = 10  # seconds

print(f"\nSTARTING NOISE DETECTION SCRIPT:")
print(f"\nConfiguration parameters: {CONFIG}")
print(f"threshold: {threshold}")
print(f"\nportion_length_in_secs: {portion_length_in_secs} secs")

print(f"\nDuomenų aplankas: {Duomenu_aplankas}")
print(f"MODEL_DIR {MODEL_DIR}")
print("Filtravimo parametrai:", fp)

print(f"\nModel file name: {model_file_name}")

# Load the trained U-Net model in Keras format
model_path = os.path.join(MODEL_DIR, model_file_name) # Path to the trained model
print(f"\nTrying to load model from: {model_path}")
try:
    model = keras.models.load_model(model_path)
except Exception as e:
    raise ValueError(f"Failed to load model from {model_path}: {e}")

if model is None:
    raise ValueError(f"Model loading returned None. Check if the file exists and is a valid Keras model: {model_path}")

# Verify model input shape
if not hasattr(model, "input_shape") or model.input_shape is None:
    raise ValueError(f"Loaded model does not have a valid input_shape attribute. Model: {model}")

expected_input_shape = model.input_shape[1:]  # e.g., (1024, 1)
if expected_input_shape != (segment_length, 1):
    raise ValueError(f"Model expects input shape {expected_input_shape}, but script uses ({segment_length}, 1)")


        #   DUOMENYS

# Švarūs įrašai
fileNames = ['1102_1.npy', '1001_8.npy', '1005_4.npy', '1006_1.npy', '1001_6.npy']

# Rodomi įrašai
fileNames = ['1002_2.npy'] # Išskirčių:0, rspragų:0, kitų triukšmų:12 Triukšmų kiekis: 23.20%
fileNames = ['1008_22.npy'] # Gera kokybė Triukšmų kiekis: 89.60%
fileNames = ['1009_16.npy'] # 86.40% Triukšmų kiekis: 86.40%
fileNames = ['1031_18.npy'] # 80.19%% Didžioji įrašo dalis be kūno
fileNames = ['1001_1.npy'] # Išskirčių:0, rspragų:0, kitų triukšmų:12 Triukšmų kiekis: 22.40%

fileNames = ['1009_12.npy'] # Quality 0, Idealus, Triukšmų kiekis: 92.80%

# start1

# Žiūrimas įrašas
fileNames = ['1010_0.npy'] # 2222, geras, tik keletas vietų su triukšmais
fileNames = ['1001_8.npy'] # s
fileNames = ['1002_1.npy'] # s
fileNames = ['1026_0.npy'] # s
fileNames = ['1097_0.npy'] # s
fileNames = ['1015_63.npy'] # s


# ++++++++++++++++++++++++++++++ Testavimui
# Duomenų aplankas su zive ir mit duomenimis
# Duomenu_aplankas = Path.home() / 'DI/ZIVEO_2025/DUOMENYS_UPD'

# Aplankas su EKG įrašais (.npy) ir anotacijomis (.json)
# db_folder = 'noisy_ecg_for_test'  # zive
# db_folder = 'clean_ecg_for_test'  # zive
# rec_dir = Path(Duomenu_aplankas, db_folder)

# fileNames = ['1001_8_noised.npy']
# fileNames = ['1001_8.npy']
# ++++++++++++++++++++++++++++++++++++++++++++++++++++++++


for fileName in fileNames:
    
    # Load the ECG data from .npy file
    filePath = os.path.join(rec_dir, fileName)             
    print(f"\nFailas: {filePath}")
    print("Loading ECG data from ecg_record.npy")
    try:
        ecg_data_orig = get_ecg_signal(filePath)
        
         # Apply ecg filter
        ecg_data = ecg_filter(ecg_data_orig, fp)
        print(f"Lowpas filter applied, filtered signal length: {len(ecg_data)}")
    
        # Load corresponding JSON file
        json_path = os.path.splitext(filePath)[0] + '.json'
        noise_indices_annotated = get_ecg_noise_indices_annotated(json_path) 
        print(f"ECG data loaded, shape: {ecg_data.shape}")
    except Exception as e:
        raise RuntimeError("Failed to load ecg_record.npy") from e


    print(f"\nFILE:")  
    print(f"filePath: {filePath}")  
    print(f"Total signal length: {len(ecg_data)} samples ({len(ecg_data)/fs:.1f} secs)")
    # Surandami pažymėti triukšmai
    print(f"Noisy intervals annotated: {len(noise_indices_annotated)}")
    print(f"Noisy indices annotated (start, end): {noise_indices_annotated}")
    noise_indices_annotated_secs = [(x/fs, y/fs) for x, y in noise_indices_annotated]  # Convert to seconds
    print(f"Noisy indices annotated secs (start, end): {noise_indices_annotated_secs}")

    # Detektuojami triukšmai

    print("\nNOISE DETECTION:")
    print(f"Model file name: {model_file_name}")
    print(f"Configuration parameters: {CONFIG}")

    noisy_indices = []
    noisy_indices_secs = []
    for i,threshold in enumerate(thresholds):
        denoised_signal, noisy_indices = run_ecg_denoising_pipeline(ecg_data, model, CONFIG, threshold)
        print("\n========================================")
        print(f"threshold: {threshold}")
        print(f"Noisy intervals detected: {len(noisy_indices)}")
        print(f"Noisy indices: {noisy_indices}")
        noisy_indices_secs = [(x/fs, y/fs) for x, y in noisy_indices]  # Convert to seconds
        print(f"Noisy indices secs: {noisy_indices_secs}")

        # Compute noise percentage
        noise_range = sum(end - start + 1 for start, end in noisy_indices)
        percentage_covered = (noise_range / len(ecg_data)) * 100
        print(f"Noise percentage: {percentage_covered:.2f}%")
        
    # Ploting entire signal with all found distortions: outliers, rdropouts and oscillations
    flag_plot_start_with_distortions_entire = True
    portion_length = portion_length_in_secs * fs
        
    if flag_plot_start_with_distortions_entire:

        # bendras aplankas grafikams
        plot_save_dir = "PLOTTINGS_4"

        # Extract the numerical part from the fileName
        base_name = os.path.basename(fileName)

        flag_plot_save = False

        addition_to_name = 'orig'
        if flag_plot_save:
            plot_dir = os.path.join(plot_save_dir, base_name + '_plot_' + addition_to_name)
            if not os.path.exists(plot_dir):
                os.makedirs(plot_dir)
        else:
            plot_dir = None

        print(f"\nplot_dir: {plot_dir}")
        
        print(f"filtered_signal length: {len(ecg_data)} ({len(ecg_data)/fs:.1f} secs)")
        print(f"portion_length: {portion_length} ({portion_length_in_secs} secs)")
        print(f"portion_length_in_secs: {portion_length_in_secs} ({portion_length} samples)")
        print(f"fs: {fs} ({1/fs:.1f} secs)")


        # Sudalijame  ecg_signal_start į fragmentus
        show_frag_indices = divide_signal_into_fragments(ecg_data, portion_length)
        print()
        print(f"show_frag_indices: {show_frag_indices}")
        show_frag_indices_secs = [(x/fs, math.floor(y/fs)) for x, y in show_frag_indices] # indexes
        print(f"show_frag_indices_secs: {show_frag_indices_secs}")
            
        num_fragment = 1
        for (show_frag_start_secs, show_frag_end_secs) in show_frag_indices_secs:
            print(f"\nFRAGMENT NR. {num_fragment}")
            print(f"plot_signal_from_in_secs: {show_frag_start_secs}")
            print(f"plot_signal_to_in_secs: {show_frag_end_secs}")
            if (show_frag_end_secs - show_frag_start_secs) > 0.5:
                plot_signal_write_plot_R_P_1(fileName, ecg_data, fs, num_fragment,
                    show_frag_start_secs, show_frag_end_secs, portion_length_in_secs,
                        plot_dir, addition_to_name,recID=None,
                        gap1_indices_secs = noisy_indices_secs,  # red
                        gap2_indices_secs = [], # blue
                        gap3_indices_secs=noise_indices_annotated_secs, # yellow
                        # gap3_indices_secs=[],
                        mark_indices_secs=[],
                        # annot_df=annot_df,
                        annot_df=pd.DataFrame(),   # tuščias df,
                        rpeak_indices_secs=[], ppeak_indices_secs=[],
                        flag_secs=True)

            num_fragment += 1

```
