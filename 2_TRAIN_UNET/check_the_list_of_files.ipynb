```python
"""
This script checks whether all data files listed in the text file
`Irasai_modeliui_1024_0_5_3_7_0_XX.txt` exist in the folder `DATA_FOR_TRAINING`,
and also verifies that each data file has its accompanying JSON metadata file.

What the script does:
1. Defines the path to the data folder and the input text file.
2. Reads all requested data filenames from the text file.
3. Scans the `DATA_FOR_TRAINING` folder and collects all existing file names.
4. For each requested data file:
   - checks whether the data file itself exists,
   - builds the expected JSON filename with the same basename,
   - checks whether that JSON file exists.
5. Prints a summary:
   - total number of files listed,
   - number of complete pairs found (data file + JSON),
   - number of missing data files,
   - number of missing JSON files.
6. Prints a detailed line-by-line report for every requested record.
7. Saves three output text files:
   - `missing_data_files.txt`       : names of missing data files,
   - `missing_json_files.txt`       : names of missing JSON files,
   - `file_json_check_report.txt`   : full detailed report.

Purpose:
The script helps verify that the training dataset is complete before further
processing or model training.
""" 

from pathlib import Path

# --------------------------------------------------
# SET PATHS
# --------------------------------------------------
DATA_FOR_TRAINING = Path("../DATA_FOR_TRAINING")
LIST_FILE = Path("Irasai_modeliui_1024_0_5_3_7_0_XX.txt")

# --------------------------------------------------
# READ FILE LIST
# --------------------------------------------------
with LIST_FILE.open("r", encoding="utf-8") as f:
    requested_files = [line.strip() for line in f if line.strip()]

print(f"Total files listed in {LIST_FILE}: {len(requested_files)}")
print()

# --------------------------------------------------
# GET ALL FILE NAMES IN DATA_FOR_TRAINING
# --------------------------------------------------
existing_files = {p.name for p in DATA_FOR_TRAINING.iterdir() if p.is_file()}
print(f"Total files in {DATA_FOR_TRAINING}: {len(existing_files)}")
print()

# --------------------------------------------------
# CHECK DATA FILES AND ACCOMPANYING JSON FILES
# --------------------------------------------------
found_pairs = []
missing_data_files = []
missing_json_files = []
missing_any = []

report_rows = []

for fname in requested_files:
    data_name = Path(fname).name
    json_name = Path(fname).with_suffix(".json").name

    data_exists = data_name in existing_files
    json_exists = json_name in existing_files

    if data_exists and json_exists:
        found_pairs.append(data_name)
    else:
        missing_any.append(data_name)
        if not data_exists:
            missing_data_files.append(data_name)
        if not json_exists:
            missing_json_files.append(json_name)

    report_rows.append({
        "data_file": data_name,
        "json_file": json_name,
        "data_status": "OK" if data_exists else "MISSING",
        "json_status": "OK" if json_exists else "MISSING",
    })

# --------------------------------------------------
# PRINT RESULTS
# --------------------------------------------------
print(f"Folder checked:              {DATA_FOR_TRAINING}")
print(f"List file:                   {LIST_FILE}")
print(f"Files in list:               {len(requested_files)}")
print(f"Full pairs found:            {len(found_pairs)}")
print(f"Records with anything missing: {len(missing_any)}")
print(f"Missing data files:          {len(missing_data_files)}")
print(f"Missing json files:          {len(missing_json_files)}")
print()

print("Detailed check:")
for row in report_rows:
    print(
        f"{row['data_file']:<30} "
        f"data: {row['data_status']:<8} "
        f"json: {row['json_status']:<8} "
        f"({row['json_file']})"
    )

# --------------------------------------------------
# PRINT ONLY MISSING SECTIONS
# --------------------------------------------------
if missing_data_files:
    print("\nMissing data files:")
    for fname in missing_data_files:
        print(fname)

if missing_json_files:
    print("\nMissing accompanying json files:")
    for fname in missing_json_files:
        print(fname)

if not missing_data_files and not missing_json_files:
    print("\nAll listed data files and their accompanying json files exist.")

# --------------------------------------------------
# SAVE RESULTS if flag_save_results is True
# --------------------------------------------------
flag_save_results = False  # Set to True to save results to text files

if (flag_save_results == True):
    out_missing_data = LIST_FILE.parent / "missing_data_files.txt"
    with out_missing_data.open("w", encoding="utf-8") as f:
        for fname in missing_data_files:
            f.write(fname + "\n")

    out_missing_json = LIST_FILE.parent / "missing_json_files.txt"
    with out_missing_json.open("w", encoding="utf-8") as f:
        for fname in missing_json_files:
            f.write(fname + "\n")

    out_report = LIST_FILE.parent / "file_json_check_report.txt"
    with out_report.open("w", encoding="utf-8") as f:
        f.write(f"Folder checked: {DATA_FOR_TRAINING}\n")
        f.write(f"List file:      {LIST_FILE}\n")
        f.write(f"Files in list:  {len(requested_files)}\n")
        f.write(f"Full pairs found: {len(found_pairs)}\n")
        f.write(f"Missing data files: {len(missing_data_files)}\n")
        f.write(f"Missing json files: {len(missing_json_files)}\n\n")

        f.write("Detailed check:\n")
        for row in report_rows:
            f.write(
                f"{row['data_file']:<30} "
                f"data: {row['data_status']:<8} "
                f"json: {row['json_status']:<8} "
                f"({row['json_file']})\n"
            )

    print(f"\nMissing data-file list saved to: {out_missing_data}")
    print(f"Missing json-file list saved to: {out_missing_json}")
    print(f"Detailed report saved to:        {out_report}")
```
